Live logging of traffic data with Python/TraCI

```python
import traci

stop_durations = {}

while traci.simulation.getMinExpectedNumber() > 0:
    traci.simulationStep()
    for veh_id in traci.vehicle.getIDList():
        speed = traci.vehicle.getSpeed(veh_id)
        if speed < 0.1:  # Steht das Fahrzeug?
            lane = traci.vehicle.getLaneID(veh_id)
            stop_durations.setdefault(lane, 0)
            stop_durations[lane] += 1  # t in sec
```

Hotspot Recognition 

```python
hotspots = {}
threshold = 30  # e.g. stop time higher than 30sec

for lane, duration in stop_durations.items():
    if duration >= threshold:
        hotspots[lane] = duration
```
