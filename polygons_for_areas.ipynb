## Draw Information from Points of Interests (POI)

When paying attention to the times of arrival and departure of agents, an assumption can be made, that certain places have certain values. I.e. an agent that departs from point A at 07:00 and arrives again at 17:00, it is relatively safe to say, that point A is the agents residence. While point B (where the agent resides from 8:00 to 16:00) is the working area.

### OSMNX

A library in python is capable of extracting data from Open Street Map and assort them to certain categories. So, pulling areas from "Stuttgart, Germany" with a tag 'residential' can retrieve such areas.

```python
import osmnx as ox

# load region
gdf = ox.geometries_from_place("Stuttgart, Germany", tags={"landuse": True})

# load only residential areas
residential_areas = gdf[gdf['landuse'] == 'residential']

# access array of residential areas' geometry
for poly in residential_areas['geometry']:
    print(poly)
```

This can also be done through a query

```python
query = """
[out:json][timeout:25];
area[name="Stuttgart"]->.searchArea;
(
  way["landuse"="residential"](area.searchArea);
);
out body;
>;
out skel qt;
"""

gdf = ox.geometries_from_overpass(query)
```
